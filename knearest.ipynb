```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score,classification_report
from sklearn.neighbors import KNeighborsClassifier
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
import time
```

```python
data=fetch_openml("fashion-MNIST",version=1,as_frame=True)
```

```python
data
```

```python
x=data.data
y=data.target.astype(int)
```

```python
x_subset,_,y_subset,_=train_test_split(
    x,y,
    train_size=12000,
    stratify=y,
    random_state=42)
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x_subset,
    y_subset,
    test_size=2000,
    stratify=y_subset,
    random_state=42)
```

```python
x_train=x_train/255.0
x_test=x_test/255.0
```

```python
values=[1,3,5,7,9,15]
results=[]
for k in values:
    knn=KNeighborsClassifier(
        n_neighbors=k,
        metric='euclidean',
        n_jobs=1)
    knn.fit(x_train,y_train)
    start_time=time.time()
    y_pred=knn.predict(x_test)
    estimated_time=time.time()-start_time
    accuracy=accuracy_score(y_test,y_pred)
    results.append({
        "k":k,
        "elapsed_time":estimated_time,
        "accuracy":accuracy,
        "prediction":y_pred})
result_table=pd.DataFrame(results)    
    
    
```

```python
result_table.columns
```

```python
best_row=result_table.loc[result_table["accuracy"].idxmax()]
```

```python
best_k=best_row["k"]
best_k
best_prediction=best_row["prediction"]
```

```python

class_names = [
    "T-shirt/top",
    "Trouser",
    "pullover",
    "dress",
    "coat",
    "sandal",
    "shirt",
    "sneaker",
    "bag",
    "ankle boot"
]
print(classification_report(
    y_test,best_prediction,target_names=class_names))
```
